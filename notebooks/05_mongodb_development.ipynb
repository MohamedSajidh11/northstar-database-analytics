```python
import pandas as pd

deliveries = pd.read_csv(
'/content/northstar_data/northstar_dataset/deliveries.csv'
)

deliveries.head()
```

```python
db = client["northstar_database"]

deliveries_collection = db["deliveries"]

print("Collection Created")
```

```python
delivery_data = deliveries.to_dict(orient='records')

print(delivery_data[0])
```

```python
deliveries_collection.insert_one(
    delivery_data[0]
)

print("ONE document inserted successfully")
```

```python
deliveries_collection.find_one()
```

```python
deliveries_collection.insert_many(
    delivery_data[1:11]
)

print("10 documents inserted successfully")
```

```python
list(
    deliveries_collection.find().limit(5)
)
```

```python
deliveries_collection.update_one(
    {"delivery_status": "Failed"},
    {
        "$set": {
            "priority_review": "Yes"
        }
    }
)

print("Document Updated")
```

```python
deliveries_collection.find_one(
    {"priority_review": "Yes"}
)
```

```python
deliveries_collection.delete_one(
    {"priority_review": "Yes"}
)

print("Document Deleted")
```

```python
pipeline = [
    {
        "$group": {
            "_id": "$delivery_status",
            "total_deliveries": {
                "$sum": 1
            }
        }
    }
]

result = list(
    deliveries_collection.aggregate(pipeline)
)

print(result)
```

## Aggregation Pipeline Insight

MongoDB aggregation pipelines were used to analyse operational delivery performance by grouping delivery records according to delivery status. Aggregation processing enables efficient large-scale analytical summarisation within NoSQL environments and supports operational decision-making for logistics management.
